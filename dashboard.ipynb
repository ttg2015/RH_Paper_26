# Paper26 pipeline dashboard
Live view of the launch bay. Reads `runs/status/status.json` (written by `python -m launch status`). **This notebook never submits jobs**; submitting stays on the command line (`python -m launch run ...`).

Colours: 🟩 done · 🟨 running/queued · 🟥 failed · 🟧 stale/legacy · ⬜ not started

```python
%matplotlib inline
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd()))
import launch
from launch import dashboard

YEARS = None   # e.g. '1980-1989' to look at one decade
```

## Everything in one panel
Refresh reruns `launch status` (one `qstat` call).

```python
dashboard.app(YEARS)
```

## Static views (for screenshots and slides)

```python
data = dashboard.load()
dashboard.draw_map(data);
```

```python
dashboard.heatmap(data, 'c1');
```

```python
for j in dashboard.failed_jobs(data):
    print(j['tag'], j['detail'])
    print(dashboard.log_tail(j['log'], 10))
```
