# Stage f checks: is the entraining plume plausible?
Reads the `Ztop/` files written by `pipeline/f.py` (plus the c2/d files for the soundings). Writes nothing.

**The question:** for a reasonable entrainment rate ε, do plumes lifted from the lowest pressure level of extreme-heat profiles top out in the mid troposphere (400–600 hPa)?

Categories per event and ε: **pass** (top in 400–600 hPa) · **too shallow** (top below 600) · **too deep** (above 400) · **no buoyancy** (plume never exceeds h*: the stable, δ < 0 events).

```python
%matplotlib inline
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd()))
import stage_f_checks as F

print(F.available())
```

## 1. Sanity check: “no buoyancy” is the stable events
At small ε the no-buoyancy share should match the share of events with δ = h_LPL − h*₅₀₀ < 0. A small gap is expected, because h* is not always lowest exactly at 500 hPa.

```python
for year, event in F.available()[:2]:
    print(F.delta_check(year, event))
```

## 2. Category fractions vs ε (Figure A prototype)
Stacked shares of each category; the black line is **pass / buoyant**, i.e. the plausibility rate among events where a plume can rise at all. Pooled over every year available.

```python
by_event = {}
for year, event in F.available():
    by_event.setdefault(event, []).append(F.open_ztop(year, event))
F.fig_fractions(by_event);
print(F.table(by_event))
```

Same, with cos(latitude) weights (E02), so high-latitude cells don't count extra:

```python
F.fig_fractions(by_event, weighted=True);
```

The green band is the 400–600 hPa window, and the dotted line is the LCL of LPL air. Change `seed` for new examples.

## 5. Per-event ε window (Figure B prototype)
For each event, the ε that puts its top at 400 hPa and at 600 hPa. A width near zero means the plume loses buoyancy all at once rather than its top sinking through the window (the “collapse” seen in testing).

```python
F.fig_windows(F.open_ztop(year, event));
```
